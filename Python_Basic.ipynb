### Sequence sum

```python
#Example -> unique pairs
for i in range(1,5):
  for j in range(1,5):
    print(i,j)
```

```python
#pattern
n=int(input('Enter the number of rows: '))
for i in range(1,n+1):
  for j in range(1,i+1):
    print("*",end='')

  print()
```

### Pattern 2

1<br>
121<br>
12321<br>
1234321<br>

```python
rows=int(input('Enter the number of rows: '))
for i in range(1,rows+1):
  for j in range(1,i+1):
    print(j,end='')
  for k in range(i-1,0,-1):
    print(k,end='')

  print()
```

# Loop Control Statement

- Break
- Continue
- Pass

```python
for i in range(1,10):
  if i==5:
    break
  print(i)
```

```python
lower= int(input('Enter lower range:'))
upper= int(input('Enter upper range:'))
for i in range(lower,upper+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
    print(i)

```

```python
#continue
for i in range(1,10):
    if i==5:
      continue
    print(i)
```

```python
#pass
for i in range(1,10):
  pass
```
