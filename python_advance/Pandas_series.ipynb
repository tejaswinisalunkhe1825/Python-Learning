### What is Pandas

Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool,

### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

### Importing Pandas

```python
import numpy as np
import pandas as pd
```

### Series from lists

```python
#string
country=['India','Pakistan','USA','Nepal','Shrilanka']
pd.Series(country)
```

```python
#integer
runs=[13,14,24,50,74,100]
pd.Series(runs)
```

```python
#custom index
marks=[67,57,89,100]
subject=['maths','english','science','hindi']
pd.Series(marks,index=subject)
```

```python
#setting a name
marks=pd.Series(marks,index=subject,name='My marks')
marks
```

### Series from dict

```python
marks={
    'maths':78,
    'english':87,
    'marathi':100,
    'science':90
}
marks_series=pd.Series(marks,name='My Marks')
marks_series
```
